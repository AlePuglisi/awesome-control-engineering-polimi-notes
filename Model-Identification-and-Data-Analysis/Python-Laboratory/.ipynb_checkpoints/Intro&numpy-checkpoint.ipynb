### Format print

```python
num = 12
name = 'Sam'
print('My number is: {one}, and my name is: {two}'.format(one=num, two=name))
```

```python
print('My number is: {}, and my name is: {}'.format(num,name))
```

```python
print('My number is: %i, and my name is: %s'%(num,name))
```

## if, elif, else Statements

# for Loops

```python
for i in range(5): #print number from 0 to 4, so in the range
    print(i)
```

```python
list_of_el = ['a', 'b', 'c'] #to access the element of a list
for el in list_of_el:
    print(el)
```

## while Loops

```python
i = 1
while i < 5:
    print(i)
    i += 1
```

## Operation

```python
vect1 = np.array([2, 5, 3, 8])
```

```python
vect2 = np.array([4, 3, 8, 1])
```

```python
vect1 + vect2 # sum element per element
```

```python
vect1 - vect2 #subtraction element wise 
```

```python
vect1 * vect2 #product element wise
```

```python
vect1 / vect2 # division element wise
```

```python
np.dot(vect1, vect2) # vector product 'dot' (liear algebra)
```

```python
vect1 **2 #square element wise
```
