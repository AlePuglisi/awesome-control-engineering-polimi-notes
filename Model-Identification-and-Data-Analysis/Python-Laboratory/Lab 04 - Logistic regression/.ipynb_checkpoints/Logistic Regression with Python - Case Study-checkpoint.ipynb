## Data Cleaning
We want to fill in missing age data instead of just dropping the missing age data rows. One way to do this is by filling in the mean age of all the passengers (imputation).
However we can adopt a smarter approach and check the average age by passenger class.

```python
plt.figure(figsize=(12, 7))
sns.boxplot(x='Pass. Class', y='Age', data=data, palette='rainbow')
```

We can see the wealthier passengers in the higher classes tend to be older, which makes sense. We'll use these average age values to impute based on Pass. Class for Age.

```python
def impute_age(cols):
    Age = cols[0]
    Pclass = cols[1]
    
    if pd.isnull(Age):

        if Pclass == 1:
            return 37

        elif Pclass == 2:
            return 29

        else:
            return 24

    else:
        return Age
```

```python
data['Age'] = data[['Age','Pass. Class']].apply(impute_age, axis=1)
```

### Drop Columns

```python
# Drop non-numeric columns
data.drop(['Ticket', 'Name'],axis=1,inplace=True)
```

# Building a Logistic Regression model

Let's start by splitting our data into a training set and test set.

## Train-Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
X = data.drop('Survived',axis=1)
y = data['Survived']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=101)
```

## Training and Predicting

```python
from sklearn.linear_model import LogisticRegression                      # Import Logistic Regression
```

```python
logmodel = LogisticRegression()                                          # Logistic Regression model instantiation
logmodel.fit(X_train, y_train)                                           # Model fitting
```

```python
predictions = logmodel.predict(X_test)                               # Predictions on test set
```

## Performance Evaluation

```python
from sklearn.metrics import confusion_matrix, classification_report
```

Confusion Matrix:
$$
\begin{array}{ccc}
& \text{Observed values} &\\
\text{Predicted values} &
\begin{bmatrix}
    TP & FP \\
    FN & TN\\
\end{bmatrix} &
\end{array}
$$ 

F1-score (harmonic mean of precision and recall):
$$
F1 = \frac{2 \times precision \times recall}{precision + recall}
$$
where:
- precision: $\frac{TP}{TP+FP}$ ratio of true positive (real survivors) over all predicted positive (all predicted as survivors, even if some are not)
- recall: $\frac{TP}{TP+FN}$ ratio of true positive (real survivors) over all that should have been predicted as survivors (all predicted as survivors, even if some are not)

```python
from sklearn.metrics import confusion_matrix
#confusion_matrix(y_test, predictions)
print(np.transpose(confusion_matrix(y_test, predictions)))
```

```python
from sklearn.metrics import plot_confusion_matrix, ConfusionMatrixDisplay

cm = np.transpose(confusion_matrix(y_test, predictions))
color = 'white'
#plt.subplot(111)

#matrix = plot_confusion_matrix(logmodel, X_test, y_test, cmap=plt.cm.Blues)
#matrix.ax_.set_title('Confusion Matrix', color=color)
matrix = ConfusionMatrixDisplay(np.transpose(confusion_matrix(y_test, predictions)),
                       display_labels=[1,0]).plot(values_format=".0f")
plt.ylabel('Predicted Label', color=color)
plt.xlabel('True Label', color=color)
plt.gcf().axes[0].tick_params(colors=color)
plt.gcf().axes[1].tick_params(colors=color)
plt.show()
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, predictions))                         # Compute precision, recall, f1-score
```
