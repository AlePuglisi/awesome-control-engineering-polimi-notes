# Matplotlib

library used to plot the data (plotly is an alternative), data visualization 


```python
import matplotlib.pyplot as plt # the .pyplot is necessary to access the plot drawing class
# % symbol used to acces the prompt from the notebook 
```

```python
%matplotlib inline
```

```python
import numpy as np
```

```python
x = np.arange(0, 5, 0.5) # x coordinates vector
```

```python
y = x**2 # y coordinates vector, as x is a np array, an operation on x will be a np array too!
```

## Basic plot command

```python
plt.plot(x, y) #.plot is the plotting function of the library
```

```python
y2 = x**3
```

```python
# to plot more lines on the same figure, with different color
plt.plot(x, y, label='function 1')
plt.plot(x, y2, label='function 2')
plt.legend() # to see the written label
```

```python
plt.plot(x, y, 'r--') #to customize the plot as red and with dashed lines ---
plt.xlabel('X axis label')
plt.ylabel('Y axis label')
plt.title('Plot Title')

plt.xlim([0,5])
plt.ylim([0,25])
plt.margins(x=0) # to remove the blank space
```

## Multiple plot in same figure

```python
plt.subplot(1, 2, 1) #(rows, columns, index of rferring plot)
plt.plot(x, y, 'r.-')
plt.xlabel('X axis label')
plt.ylabel('Y axis label')
plt.title('Plot Title 1')
plt.subplot(1, 2, 2)
plt.plot(y, x, 'g*-')
plt.xlabel('X axis label')
plt.ylabel('Y axis label')
plt.title('Plot Title 2')
```

## Size, DPI, dimension

```python
plt.figure(figsize=(8,4), dpi=100) #high dpi to have a good resolution, save it good
plt.plot(x, y, 'r--', label='function 1')
plt.plot(x, y2, 'b.-', label='function 2')
plt.legend() 
#to remove the small text before the image:
plt.grid()
plt.show()
```

 ## Saving figure

```python
plt.figure(figsize=(8,4), dpi=100) #high dpi to have a good resolution, save it good
plt.plot(x, y, 'r--', label='function 1')
plt.plot(x, y2, 'b.-', label='function 2')
plt.legend() 
#to remove the small text before the image:
plt.grid()
plt.show()
plt.savefig('figure.png')
```

## Scatter plot

```python
plt.scatter(x, y) #to look at the points, without interpolating
```

```python
# histogram
from random import sample # generate random samples
data = sample(range(1, 1000), 100) #from 1 to 1000 choose a sample 100 times
plt.hist(data, 30);
```

```python
# boxplot
data = [np.random.normal(0, std, 100) for std in range(1,4)] #from a normal distribution, created with list comprehension
plt.boxplot(data, vert=True);
# representing the mean and the standard deviation into a graph
```

## Selection and indexing

```python
df['W'] # get access to column W like a list, which is a pandas series datatype
```

```python
df[['W', 'Z']] #look into more columns at a time
```

## New column

```python
df['new'] = df['W'] + df['Y'] # new column containing sum of column W and Y
df
```

## Remove column/row

```python
df.drop('new', axis=1) #axis=0 means 'row' | axis = 1 means 'column'
```

```python
#NB some operation modify directly the dataframe! infact i have top assign the new df with the dropped column
df = df.drop('new', axis=1)
```

```python
#i can remove it inoplace
#df.drop('new', axis=1, inplace=True)
```

```python
df.drop('E', axis=0, inplace=True)
df
```

## Selecting rows

## Conditional Selection

```python
df[df['W']<0] #take all the elements less than 0
```

```python
df[df['W']<0]['Y'] #look at all element less than 0, looking only Y column
```

```python
df[(df['W']<0)&(df['Y']>-0.5)] #evaluate multiple condition (each one between round brackets)
```
